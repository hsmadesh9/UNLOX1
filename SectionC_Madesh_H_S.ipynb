```python
def q16_swap():
    print("Q16 - Swap two numbers")
    a = int(input("Enter a: "))
    b = int(input("Enter b: "))

    # Method 1: arithmetic (for numeric values)
    x, y = a, b
    x = x + y
    y = x - y
    x = x - y
    print("Arithmetic:", x, y)




def q17_order_calculator():
    print("Q17 - Swiggy order calculator")
    price = float(input("Enter item price (₹): "))
    quantity = int(input("Enter quantity: "))
    subtotal = price * quantity
    gst = subtotal * 0.18
    delivery = 40 if subtotal < 500 else 0
    total = subtotal + gst + delivery
    print(f"Subtotal:     ₹{subtotal:.2f}")
    print(f"GST (18%):    ₹{gst:.2f}")
    print(f"Delivery:     ₹{delivery:.2f}")
    print("------------------------")
    print(f"Total:        ₹{total:.2f}")




def q18_temperature_converter():
    print("Q18 - Temperature converter")
    value = float(input("Enter temperature value: "))
    unit = input("Enter unit (C/F): ").strip().upper()
    if unit == "C":
        result = (value * 9 / 5) + 32
        print(f"{value:.2f} °C = {result:.2f} °F")
    elif unit == "F":
        result = (value - 32) * 5 / 9
        print(f"{value:.2f} °F = {result:.2f} °C")
    else:
        print("Error: invalid unit. Please enter C or F.")


def q19_sum_four_digits():
    print("Q19 - Sum of digits of a 4-digit number")
    number = int(input("Enter a 4-digit positive number: "))
    if number < 1000 or number > 9999:
        print("Error: enter a number from 1000 to 9999.")
        return
    n = number
    d1 = n % 10
    n = n // 10
    d2 = n % 10
    n = n // 10
    d3 = n % 10
    n = n // 10
    d4 = n % 10
    print("Sum of digits:", d1 + d2 + d3 + d4)



def q20_smart_input_parser():
    print("Q20 - Smart input parser")
    raw = input("Enter a value: ")
    try:
        value = int(raw)
        detected_type = "int"
    except ValueError:
        try:
            value = float(raw)
            detected_type = "float"
        except ValueError:
            value = raw
            detected_type = "str"
    print("Value:", value)
    print("Detected type:", detected_type)






def q21_income_tax():
    print("Q21 - Slab-wise income tax calculator (slabs exactly as in assignment)")
    income = float(input("Enter annual income (₹): "))
    if income < 0:
        print("Income cannot be negative.")
        return

    remaining = income
    tax = 0.0
    slabs = [
        (300000, 0.00, "Up to ₹3,00,000"),
        (400000, 0.05, "Next ₹4,00,000 (₹3L-₹7L)"),
        (300000, 0.10, "Next ₹3,00,000 (₹7L-₹10L)"),
        (200000, 0.15, "Next ₹2,00,000 (₹10L-₹12L)"),
        (300000, 0.20, "Next ₹3,00,000 (₹12L-₹15L)"),
        (float("inf"), 0.30, "Above ₹15,00,000"),
    ]
    print("Tax breakdown:")
    for width, rate, label in slabs:
        taxable = min(remaining, width)
        slab_tax = taxable * rate
        print(f"{label}: taxable ₹{taxable:,.2f} at {rate*100:.0f}% = ₹{slab_tax:,.2f}")
        tax += slab_tax
        remaining -= taxable
        if remaining <= 0:
            break
    print(f"Total tax (before any cess/rebate): ₹{tax:,.2f}")





    matrix = np.eye(5)
    print("Identity matrix:\\n", matrix)
    matrix = matrix * 10
    print("Multiplied by 10:\\n", matrix)
    transposed = matrix.T
    # An identity matrix is symmetric; its transpose equals itself.
    print("Transpose:\\n", transposed)
    print("Diagonal sum:", np.trace(transposed))
    print("Sum of all elements:", np.sum(transposed))



def q22_pan_validator():
    print("Q22 - PAN format validator")
    pan = input("Enter PAN: ").strip()
    if len(pan) != 10:
        print("Invalid PAN — length must be exactly 10 characters.")
    elif not all("A" <= ch <= "Z" for ch in pan[:5]):
        print("Invalid PAN — first five characters must be uppercase letters.")
    elif not all("0" <= ch <= "9" for ch in pan[5:9]):
        print("Invalid PAN — characters 6 to 9 must be digits.")
    elif not ("A" <= pan[9] <= "Z"):
        print("Invalid PAN — last character must be an uppercase letter.")
    else:
        print("Valid PAN")


def q23_bescom_bill():
    print("Q23 - BESCOM residential bill (rates supplied in assignment)")
    units = float(input("Enter units consumed: "))
    if units < 0:
        print("Units cannot be negative.")
        return
    # Progressive slab calculation, charging each block at its stated rate.
    remaining = units
    energy = 0.0
    slabs = [(50, 4.10), (50, 5.55), (100, 7.10), (float("inf"), 8.15)]
    for width, rate in slabs:
        used = min(remaining, width)
        energy += used * rate
        remaining -= used
        if remaining <= 0:
            break
    fixed = 110.0
    tax = (energy + fixed) * 0.06
    total = energy + fixed + tax
    print(f"Units consumed:       {units:.2f}")
    print(f"Energy charges:       ₹{energy:.2f}")
    print(f"Fixed charge:         ₹{fixed:.2f}")
    print(f"Government tax (6%):  ₹{tax:.2f}")
    print("------------------------------")
    print(f"Total bill:           ₹{total:.2f}")


def q24_cricket_strike_rate():
    print("Q24 - Cricket strike-rate classifier")
    runs = float(input("Enter runs scored: "))
    balls = float(input("Enter balls faced: "))
    if balls == 0:
        print("Cannot calculate strike rate: balls faced is zero.")
        return
    sr = (runs / balls) * 100
    if sr > 150:
        category = "Explosive (T20 finisher)"
    elif sr > 100:
        category = "Aggressive (Top order T20)"
    elif sr >= 75:
        category = "Steady (ODI anchor)"
    else:
        category = "Defensive (Test match)"
    print(f"Strike rate: {sr:.2f}")
    print("Category:", category)




def q25_bmi():
    print("Q25 - BMI calculator using assignment's Indian categories")
    weight = float(input("Enter weight in kg: "))
    height_cm = float(input("Enter height in cm: "))
    if weight <= 0 or height_cm <= 0:
        print("Weight and height must be positive.")
        return
    height_m = height_cm / 100
    bmi = weight / (height_m * height_m)
    if bmi < 18.5:
        category = "Underweight"
    elif bmi < 23:
        category = "Normal"
    elif bmi < 25:
        category = "Overweight"
    elif bmi < 30:
        category = "Obese Class I"
    else:
        category = "Obese Class II"
    print(f"BMI: {bmi:.2f}")
    print("Category:", category)


def q26_number_triangle():
    print("Q26 - Number triangle")
    n = int(input("Enter N: "))
    for row in range(1, n + 1):
        for number in range(1, row + 1):
            print(number, end=" ")
        print()


def q27_number_diamond():
    print("Q27 - Number diamond")
    n = int(input("Enter N: "))
    # Upper half including the middle row.
    for row in range(1, n + 1):
        print(" " * (n - row), end="")
        for number in range(1, row + 1):
            print(number, end=" ")
        for number in range(row - 1, 0, -1):
            print(number, end=" ")
        print()
    # Lower half.
    for row in range(n - 1, 0, -1):
        print(" " * (n - row), end="")
        for number in range(1, row + 1):
            print(number, end=" ")
        for number in range(row - 1, 0, -1):
            print(number, end=" ")
        print()








def q28_hollow_square():
    print("Q28 - Hollow square")
    n = int(input("Enter N: "))
    for row in range(n):
        for col in range(n):
            if row == 0 or row == n - 1 or col == 0 or col == n - 1:
                print("*", end=" ")
            else:
                print(" ", end=" ")
        print()


def q29_pascals_triangle():
    print("Q29 - Pascal's triangle (first 5 rows)")
    rows = 5
    for row in range(rows):
        value = 1
        print(" " * (rows - row), end="")
        for col in range(row + 1):
            print(value, end=" ")
            value = value * (row - col) // (col + 1)
        print()


def q30_alphabet_triangle():
    print("Q30 - Alphabet triangle")
    n = int(input("Enter N (1 to 26): "))
    if n < 1 or n > 26:
        print("Enter N from 1 to 26.")
        return
    for row in range(1, n + 1):
        for col in range(row):
            print(chr(ord("A") + col), end=" ")
        print()


def q31_armstrong_numbers():
    print("Q31 - Armstrong numbers from 1 to 9999")
    found = []
    for number in range(1, 10000):
        digits_count = 0
        temp = number
        while temp > 0:
            digits_count += 1
            temp //= 10
        total = 0
        temp = number
        while temp > 0:
            digit = temp % 10
            total += digit ** digits_count
            temp //= 10
        if total == number:
            found.append(number)
    print(", ".join(str(number) for number in found))
    print("Count:", len(found))


def q32_fibonacci():
    print("Q32 - Fibonacci series and sum")
    n = int(input("Enter number of terms N: "))
    if n < 0:
        print("N cannot be negative.")
        return
    first, second = 0, 1
    total = 0
    terms = []
    for _ in range(n):
        terms.append(first)
        total += first
        first, second = second, first + second
    print("Terms:", terms)
    print("Sum:", total)


def q33_prime_and_sum():
    print("Q33 - Prime check and sum of primes up to N")
    n = int(input("Enter N: "))

    def is_prime(number):
        if number < 2:
            return False
        divisor = 2
        while divisor * divisor <= number:
            if number % divisor == 0:
                return False
            divisor += 1
        return True

    print(f"{n} is {'prime' if is_prime(n) else 'not prime'}.")
    prime_sum = 0
    primes = []
    for number in range(2, n + 1):
        if is_prime(number):
            primes.append(number)
            prime_sum += number
    print("Primes:", primes)
    print("Sum of primes:", prime_sum)











def q34_reverse_palindrome_number():
    print("Q34 - Reverse number and palindrome check")
    number = int(input("Enter an integer: "))
    original = number
    negative = number < 0
    number = abs(number)
    reversed_number = 0
    while number > 0:
        digit = number % 10
        reversed_number = reversed_number * 10 + digit
        number //= 10
    display_reversed = -reversed_number if negative else reversed_number
    print("Reversed number:", display_reversed)
    if negative:
        print("Not a palindrome (negative numbers are not palindromes here).")
    else:
        print("Palindrome" if original == reversed_number else "Not a palindrome")





def q35_hcf_lcm():
    print("Q35 - HCF and LCM using Euclidean algorithm")
    a = int(input("Enter first positive integer: "))
    b = int(input("Enter second positive integer: "))
    if a <= 0 or b <= 0:
        print("Both numbers must be positive.")
        return
    original_a, original_b = a, b
    while b != 0:
        a, b = b, a % b
    hcf = a
    lcm = (original_a * original_b) // hcf
    print("HCF:", hcf)
    print("LCM:", lcm)






def q36_string_scanner():
    print("Q36 - Single-pass string scanner")
    sentence = input("Enter a sentence: ")
    total_chars = len(sentence)
    words = 0
    vowels = consonants = digits = spaces = special = 0
    in_word = False

    for ch in sentence:
        if ch.isspace():
            spaces += 1
            in_word = False
        else:
            if not in_word:
                words += 1
                in_word = True
            if ch.isalpha():
                if ch.lower() in "aeiou":
                    vowels += 1
                else:
                    consonants += 1
            elif ch.isdigit():
                digits += 1
            else:
                special += 1

    print("Total characters:", total_chars)
    print("Total words:", words)
    print("Vowels:", vowels)
    print("Consonants:", consonants)
    print("Digits:", digits)
    print("Spaces:", spaces)
    print("Special characters:", special)


def q37_marks_analysis():
    print("Q37 - Student marks analysis")
    marks = list(map(float, input("Enter 10 marks separated by spaces: ").split()))
    if len(marks) != 10:
        print("Please enter exactly 10 marks.")
        return
    maximum = minimum = marks[0]
    total = 0
    for mark in marks:
        if mark > maximum:
            maximum = mark
        if mark < minimum:
            minimum = mark
        total += mark
    average = total / len(marks)
    above_average = 0
    for mark in marks:
        if mark > average:
            above_average += 1

    # Find second-highest DISTINCT mark without sort()/sorted().
    highest = second_highest = None
    for mark in marks:
        if highest is None or mark > highest:
            second_highest = highest
            highest = mark
        elif mark != highest and (second_highest is None or mark > second_highest):
            second_highest = mark

    print("Maximum:", maximum)
    print("Minimum:", minimum)
    print("Average:", average)
    print("Students above average:", above_average)
    print("Second highest distinct mark:", second_highest)


def q38_word_frequency():
    print("Q38 - Word frequency analyzer")
    sentence = input("Enter review sentence: ")
    cleaned = ""
    for ch in sentence.lower():
        if ch.isalnum() or ch.isspace():
            cleaned += ch
        else:
            cleaned += " "
    frequency = {}
    for word in cleaned.split():
        frequency[word] = frequency.get(word, 0) + 1

    # Sort dictionary items by count descending; alphabetical order breaks ties.
    ordered = sorted(frequency.items(), key=lambda item: (-item[1], item[0]))
    for word, count in ordered:
        print(word, ":", count)
    if ordered:
        print("Most frequent:", ordered[0][0], "Count:", ordered[0][1])
    else:
        print("No words found.")


def q39_states_crud():
    print("Q39 - Indian states CRUD")
    states = {
        "Karnataka": "Bengaluru",
        "Maharashtra": "Mumbai",
        "Tamil Nadu": "Chennai",
        "Kerala": "Thiruvananthapuram",
        "Telangana": "Hyderabad",
    }
    while True:
        print("\\n1. Add  2. Update  3. Delete  4. Search  5. Display  6. Exit")
        choice = input("Choose option: ").strip()
        if choice == "1":
            state = input("State name: ").strip()
            capital = input("Capital: ").strip()
            states[state] = capital
            print("State added.")
        elif choice == "2":
            state = input("State to update: ").strip()
            if state in states:
                states[state] = input("New capital: ").strip()
                print("Capital updated.")
            else:
                print("State not found.")
        elif choice == "3":
            state = input("State to delete: ").strip()
            if state in states:
                del states[state]
                print("State deleted.")
            else:
                print("State not found.")
        elif choice == "4":
            state = input("State to search: ").strip()
            print(states.get(state, "Not found"))
        elif choice == "5":
            for state in sorted(states):
                print(state, ":", states[state])
        elif choice == "6":
            print("Exiting.")
            break
        else:
            print("Invalid option. Choose 1 to 6.")





def q40_set_operations():
    print("Q40 - Set operations")
    section_a = {"Python", "Java", "C++", "SQL"}
    section_b = {"Java", "JavaScript", "Python", "Go", "SQL"}
    print("Both sections:", section_a & section_b)
    print("Only Section A:", section_a - section_b)
    print("Only Section B:", section_b - section_a)
    print("All unique languages:", section_a | section_b)
    print("Exclusive to one section:", section_a ^ section_b)



def q41_stock_analysis():
    print("Q41 - TCS weekly stock analysis")
    prices = (3845, 3902, 3878, 3955, 4012, 3998, 4055)
    days = ("Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun")
    max_price = max(prices)
    min_price = min(prices)
    average = sum(prices) / len(prices)
    max_gain = prices[1] - prices[0]
    gain_start, gain_end = 0, 1
    volatility = 0
    for index in range(1, len(prices)):
        change = prices[index] - prices[index - 1]
        volatility += abs(change)
        if change > max_gain:
            max_gain = change
            gain_start, gain_end = index - 1, index
    print("Maximum:", max_price, "on", days[prices.index(max_price)])
    print("Minimum:", min_price, "on", days[prices.index(min_price)])
    print(f"Average: ₹{average:.2f}")
    print(f"Biggest single-day gain: ₹{max_gain} ({days[gain_start]} to {days[gain_end]})")
    print("Total volatility:", volatility)


def q42_duplicate_analysis():
    print("Q42 - Duplicate analysis")
    numbers = [4, 7, 2, 4, 9, 7, 1, 3, 2, 5, 8, 9, 6, 3, 10, 11, 5, 12, 13, 10]
    unique = set(numbers)
    repeated = []
    exactly_once = []
    for number in unique:
        count = 0
        for item in numbers:
            if item == number:
                count += 1
        if count > 1:
            repeated.append(number)
        elif count == 1:
            exactly_once.append(number)
    print("Original list:", numbers)
    print("Unique numbers (set):", unique)
    print("Appearing more than once:", sorted(repeated))
    print("Appearing exactly once:", sorted(exactly_once))





def q43_sentence_palindrome():
    print("Q43 - Sentence palindrome checker")
    sentence = input("Enter sentence: ")
    cleaned = ""
    for ch in sentence.lower():
        if ch.isalnum():
            cleaned += ch
    if cleaned == cleaned[::-1]:
        print("True - palindrome")
    else:
        print("False - not a palindrome")





def q44_manual_title_case():
    print("Q44 - Manual title case")
    sentence = input("Enter sentence: ")
    result = ""
    start_of_word = True
    for ch in sentence:
        if ch.isspace():
            result += ch
            start_of_word = True
        elif start_of_word:
            result += ch.upper()
            start_of_word = False
        else:
            result += ch.lower()
    print(result)

def q45_caesar_cipher():
    print("Q45 - Caesar cipher")
    message = input("Enter message: ")
    shift = int(input("Enter shift value: ")) % 26
    encrypted = ""
    for ch in message:
        if "A" <= ch <= "Z":
            encrypted += chr((ord(ch) - ord("A") + shift) % 26 + ord("A"))
        elif "a" <= ch <= "z":
            encrypted += chr((ord(ch) - ord("a") + shift) % 26 + ord("a"))
        else:
            encrypted += ch
    print("Encrypted message:", encrypted)






def q46_calculate_gst(amount, gst_rate=18, gst_inclusive=False):
    """Return (base amount, GST amount, total amount), rounded to 2 decimals."""
    rate = gst_rate / 100
    if gst_inclusive:
        total = amount
        base = amount / (1 + rate)
        gst = total - base
    else:
        base = amount
        gst = base * rate
        total = base + gst
    return (round(base, 2), round(gst, 2), round(total, 2))

def run_q46():
    print("Q46 - GST function")
    print("calculate_gst(1000) =", q46_calculate_gst(1000))
    print("calculate_gst(1180, gst_inclusive=True) =", q46_calculate_gst(1180, gst_inclusive=True))
    print("calculate_gst(500, gst_rate=5) =", q46_calculate_gst(500, gst_rate=5))


def q47_power(base, exp):
    """Recursive power without ** or pow()."""
    if exp == 0:
        return 1
    if exp < 0:
        if base == 0:
            raise ZeroDivisionError("Zero cannot be raised to a negative power.")
        return 1 / q47_power(base, -exp)
    return base * q47_power(base, exp - 1)


def run_q47():
    print("Q47 - Recursive power")
    print("power(2, 10) =", q47_power(2, 10))
    print("power(5, 0) =", q47_power(5, 0))
    print("power(3, -2) =", q47_power(3, -2))


def q48_numpy_identity():
    import numpy as np
    print("Q48 - NumPy identity matrix")
    matrix = np.eye(5)
    print("Identity matrix:\\n", matrix)
    matrix = matrix * 10
    print("Multiplied by 10:\\n", matrix)
    transposed = matrix.T
    # An identity matrix is symmetric; its transpose equals itself.
    print("Transpose:\\n", transposed)
    print("Diagonal sum:", np.trace(transposed))
    print("Sum of all elements:", np.sum(transposed))













def q49_numpy_sales():
    import numpy as np
    print("Q49 - Zomato weekly sales analytics")
    sales = np.array([12000, 18500, 14200, 22100, 19800, 35200, 28400])
    days = np.array(["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"])
    total = np.sum(sales)
    average = np.mean(sales)
    max_day = days[np.argmax(sales)]
    above_average_count = np.sum(sales > average)
    monday_to_sunday_growth = ((sales[-1] - sales[0]) / sales[0]) * 100
    day_changes = (sales[1:] - sales[:-1]) / sales[:-1] * 100
    print("Sales:", sales)
    print("Total weekly sales: ₹", total)
    print("Daily average: ₹", round(float(average), 2))
    print("Maximum sales day:", max_day)
    print("Days above average:", int(above_average_count))
    print(f"Monday-to-Sunday growth: {monday_to_sunday_growth:.2f}%")
    print("Day-on-day percentage changes:", np.round(day_changes, 2))


def q50_numpy_border():
    import numpy as np
    print("Q50 - NumPy border pattern (no explicit loops)")
    matrix = np.ones((5, 5), dtype=int)
    matrix[1:-1, 1:-1] = 0
    print(matrix)


q16_swap
q17_order_calculator
q18_temperature_converter
q19_sum_four_digits
q20_smart_input_parser
q21_income_tax
q22_pan_validato
q23_bescom_bill
q24_cricket_strike_rate
q25_bmi
q26_number_triangle
q27_number_diamond
q28_hollow_square
q29_pascals_triangle
q30_alphabet_triangle
q31_armstrong_numbers
q32_fibonacci
q33_prime_and_sum
q34_reverse_palindrome_number
q35_hcf_lcm
q36_string_scanner
q37_marks_analysis
q38_word_frequency
q39_states_crud
q40_set_operations
q41_stock_analysis
q42_duplicate_analysis
q43_sentence_palindrome
q44_manual_title_case
q45_caesar_cipher
run_q46
run_q47
q48_numpy_identity
q49_numpy_sales
q50_numpy_border


#all this code are done in pychaem then i past it to this becose i dont know about this colab


```
