```python
"""
Q1. `is` vs `==`
`==` checks whether two objects have equal values.
`is` checks whether two names refer to the exact same object in memory.
Example:
a = [1, 2]
b = [1, 2]
print(a == b)  # True: same contents
print(a is b)  # False: two different list objects

Q2. Mutable and immutable
Mutable: list, set, dict.
Immutable: tuple, str, int, float, bool, frozenset.
Mutable objects can be changed after creation. Immutable objects cannot be
changed; an operation that seems to modify one creates a new object instead.
Immutability helps make values predictable and safe to share.
Q3. Floating-point comparison
print(0.1 + 0.2 == 0.3)  # False
Binary floating-point cannot represent some decimal fractions exactly.
Therefore 0.1 + 0.2 is stored as a value extremely close to, but not exactly,
0.3. For real comparisons, use a tolerance such as abs(a-b) < 1e-9.

Q4. When to choose a tuple
1. Fixed coordinates, e.g. (12, 20): the pair should not be changed.
2. A function returning several fixed results: a tuple communicates a fixed
   group of values.
3. A dictionary key: a tuple containing hashable values can be a key, while
   a list cannot. Tuples also communicate that the collection is fixed.
Q5. Dictionary lookup vs list search
Without saying hash: a dictionary uses a key-based lookup structure to go
directly to the likely location of an entry; a list checks items one by one.
With the word hash: a dictionary computes a hash of the key and uses it to
locate a bucket, so lookup is O(1) on average. Searching a list for a value
may inspect all n elements, so it is O(n)
Q6. range() vs numpy.arange()
1. range() is built into Python and produces a memory-efficient range object;
   np.arange() creates a NumPy array.
2. range() is for integer sequences; np.arange() supports integer and
   floating-point steps.
3. range() excludes the stop value; np.arange() also normally excludes it,
   but floating-point steps can have precision surprises.
Example: list(range(1, 5, 1)) -> [1, 2, 3, 4]
         np.arange(0, 1, 0.2) -> approximately [0., .2, .4, .6, .8]

Q7. append(), extend(), insert()
a = [1, 2, 3]; x = [4, 5]
a.append(x) -> [1, 2, 3, [4, 5]] (adds x as one item)
a.extend(x) -> [1, 2, 3, 4, 5] (adds each item)
a.insert(1, x) -> [1, [4, 5], 2, 3] (inserts x at index 1)
Q8. Extended unpacking
a, *b, c = [10, 20, 30, 40, 50]
a = 10, b = [20, 30, 40], c = 50.
The starred variable collects the remaining middle items in a list.
[10, 20] gives a=10, b=[], c=20.
[10] raises ValueError because a and c both need a value.

Q9. Operators applied to (-7, 2)
-7 / 2  -> -3.5  (true division)
-7 // 2 -> -4    (floor division, rounds down toward negative infinity)
-7 % 2  -> 1     (remainder consistent with floor division)
-7 ** 2 -> 49    (power; the second operand is the exponent)
Q10. Why NumPy is faster for numerical work
1. NumPy arrays store same-type values in compact, usually contiguous memory,
   unlike Python lists which store references to Python objects.
2. NumPy performs many operations in optimized compiled code.
3. Vectorized operations process whole arrays without a Python-level loop.
"""
```
